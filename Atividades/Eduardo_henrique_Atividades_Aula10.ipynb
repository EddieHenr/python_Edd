# Atividade 01


```python

import json
import csv

cluster = {
    "cluster": {
        "nome": "Cluster_Principal",
        "versao_SO": "Ubuntu Server 22.04",
        "servidores": [
            {
                "servidor": "Servidor 1",
                "IP": "192.168.0.10",
                "portas": [80, 443]
            },
            {
                "servidor": "Servidor 2",
                "IP": "192.168.0.11",
                "portas": [3306, 22]
            },
            {
                "servidor": "Servidor 3",
                "IP": "192.168.0.12",
                "portas": [8080, 21]
            }
        ]
    }
}

print("Dicionário criado com sucesso!")
print(cluster)


# Etapa 2: Exportação JSON

with open("cluster_config.json", "w", encoding="utf-8") as arquivo:
    json.dump(cluster, arquivo, indent=4, ensure_ascii=False)

print("\nArquivo cluster_config.json exportado com sucesso!")


# Etapa 3: Exportação CSV

with open("servidores_cluster.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo, delimiter=";")

    escritor.writerow(["Servidor", "IP", "Portas"])

    for servidor in cluster["cluster"]["servidores"]:
        escritor.writerow([
            servidor["servidor"],
            servidor["IP"],
            ", ".join(map(str, servidor["portas"]))
        ])

print("Arquivo servidores_cluster.csv exportado com sucesso!")
```

# Atividade 02

```python

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill

# Etapa 2: Criação da planilha

planilha = Workbook()
aba = planilha.active
aba.title = "Servidores"

# Criando o cabeçalho
cabecalho = ["Hostname", "IP", "Memória (GB)", "Uso de Disco (%)", "Status"]

aba.append(cabecalho)

# Etapa 3: Inserção de dados

servidores = [
    ["Servidor01", "192.168.0.10", 16, 75, "Ativo"],
    ["Servidor02", "192.168.0.11", 32, 85, "Ativo"],
    ["Servidor03", "192.168.0.12", 8, 90, "Ativo"],
    ["Servidor04", "192.168.0.13", 16, 60, "Inativo"]
]

for servidor in servidores:
    aba.append(servidor)

# Etapa 4: Avisos de uso de disco

for linha in range(2, aba.max_row + 1):
    uso_disco = aba.cell(row=linha, column=4).value

    if uso_disco >= 80:
        aba.cell(row=linha, column=4).fill = PatternFill(
            start_color="FF0000",
            end_color="FF0000",
            fill_type="solid"
        )

        aba.cell(row=linha, column=4).font = Font(color="FFFFFF", bold=True)

        hostname = aba.cell(row=linha, column=1).value

        print(f"ALERTA: O servidor {hostname} precisa de expansão de armazenamento!")

# Ajustando o cabeçalho
for celula in aba[1]:
    celula.font = Font(bold=True)
    celula.fill = PatternFill(
        start_color="ADD8E6",
        end_color="ADD8E6",
        fill_type="solid"
    )

# Salvando a planilha
planilha.save("inventario_servidores.xlsx")

print("\nPlanilha inventario_servidores.xlsx criada com sucesso!")
```

# Atividade 03

```python

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill

# Etapa 1: Abertura do arquivo log.txt

try:
    with open("log.txt", "r", encoding="utf-8") as arquivo:
        linhas = arquivo.readlines()

    print("Arquivo log.txt aberto com sucesso!")

except FileNotFoundError:
    print("Erro: O arquivo log.txt não foi encontrado.")
    linhas = []


# Etapa 2: Extração de dados

dados_logs = []

for linha in linhas:
    linha = linha.strip()

    if linha:
        partes = linha.split(" ", 2)

        data_hora = partes[0]
        tipo_evento = "INFO"
        mensagem = linha

        if "INFO" in linha:
            tipo_evento = "INFO"

        elif "AVISO" in linha:
            tipo_evento = "AVISO"

        elif "ERRO" in linha:
            tipo_evento = "ERRO"

        dados_logs.append([data_hora, tipo_evento, mensagem])


# Etapa 3: Criação da Planilha

planilha = Workbook()
aba = planilha.active
aba.title = "Auditoria de Logs"

# Criando o cabeçalho
aba.append(["Data/Hora", "Tipo de Evento", "Mensagem"])

# Inserindo os dados
for log in dados_logs:
    aba.append(log)


# Etapa 4: Salvando a planilha

# Formatação do cabeçalho
for celula in aba[1]:
    celula.font = Font(bold=True)
    celula.fill = PatternFill(
        start_color="ADD8E6",
        end_color="ADD8E6",
        fill_type="solid"
    )

# Ajustando largura das colunas
aba.column_dimensions["A"].width = 20
aba.column_dimensions["B"].width = 20
aba.column_dimensions["C"].width = 60

# Salvando o arquivo Excel
planilha.save("auditoria_logs.xlsx")

print("\nPlanilha auditoria_logs.xlsx criada com sucesso!")
```
