# Atividade 01

```python
# ==========================================
# ETAPA 1: Inicialização do Log (Modo Escrita - w)
# ==========================================
# O modo 'w' cria o arquivo se não existir ou sobrescreve caso já exista.
with open("acessos.log", "w", encoding="utf-8") as arquivo:
    arquivo.write("=== SISTEMA DE AUDITORIA DE LOGS ===\n")
    arquivo.write("[2026-10-14 08:00:00] SISTEMA INICIALIZADO\n")

print("Etapa 1 concluída: Arquivo criado e inicializado.")


# ==========================================
# ETAPA 2: Registro de Atividades (Modo Append - a)
# ==========================================
# O modo 'a' adiciona conteúdo ao final sem apagar o que já está lá.
novos_eventos = [
    "[2026-10-14 08:05:12] LOGIN: Usuario admin acessou o sistema.\n",
    "[2026-10-14 08:15:30] ERRO: Falha na autenticacao do usuario joao.\n",
    "[2026-10-14 08:20:45] LOGIN: Usuario maria acessou o sistema.\n",
    "[2026-10-14 08:30:10] ALERTA: Tentativa de acesso nao autorizado.\n",
    "[2026-10-14 08:45:00] LOGIN: Usuario admin realizou logout.\n"
]

with open("acessos.log", "a", encoding="utf-8") as arquivo:
    arquivo.writelines(novos_eventos)

print("Etapa 2 concluída: 5 eventos adicionados com sucesso.")


# ==========================================
# ETAPA 3: Leitura e Inspeção Seletiva
# ==========================================
print("\n--- ETAPA 3: Inspeção Seletiva ---")

# 1 - Inspeção do cabeçalho usando readline()
with open("acessos.log", "r", encoding="utf-8") as arquivo:
    cabecalho = arquivo.readline()
    print(f"1. Cabeçalho: {cabecalho.strip()}")

# 2 - Último evento registrado usando readlines()
with open("acessos.log", "r", encoding="utf-8") as arquivo:
    todas_linhas = arquivo.readlines()
    if todas_linhas:
        ultimo_evento = todas_linhas[-1]
        print(f"2. Último evento: {ultimo_evento.strip()}")


# ==========================================
# ETAPA 4: Auditoria de Segurança
# ==========================================
print("\n--- ETAPA 4: Auditoria de Segurança ---")

contador_palavras = 0

# Percorrendo o arquivo linha por linha (eficiente para arquivos grandes)
with open("acessos.log", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        # Verificando se a linha contém LOGIN ou ERRO
        if "LOGIN" in linha or "ERRO" in linha:
            print(f"-> {linha.strip()}")
            contador_palavras += 1

print(f"\nContagem total (LOGIN / ERRO): {contador_palavras}")
```

# Atividade 02

```python

# Etapa 1: Tratamento de erros (FileNotFoundError)

try:
    with open("servidor_inexistente.log", "r", encoding="utf-8") as arquivo:
        conteudo = arquivo.read()
        print(conteudo)

except FileNotFoundError:
    print("Erro: O arquivo solicitado não foi encontrado.")

finally:
    print("Tentativa de leitura finalizada.")


# Leitura do arquivo principal
try:
    with open("acessos.log", "r", encoding="utf-8") as arquivo:
        conteudo = arquivo.read()
        print("\nConteúdo do arquivo acessos.log:")
        print(conteudo)

except FileNotFoundError:
    print("Erro: O arquivo acessos.log não existe.")


# Etapa 2: Proteção de escrita e permissões (PermissionError)

try:
    with open("acessos.log", "a", encoding="utf-8") as arquivo:
        arquivo.write("LOGIN: Novo acesso registrado.\n")
        print("Evento registrado com sucesso.")

except PermissionError:
    print("Erro: Permissão negada para escrever no arquivo.")


# Etapa 3: Tratamento de decodificação (UnicodeDecodeError)

try:
    with open("acessos.log", "r", encoding="ascii") as arquivo:
        conteudo = arquivo.read()
        print("\nConteúdo do arquivo:")
        print(conteudo)

except UnicodeDecodeError:
    print("Erro: O arquivo possui uma codificação incompatível.")

except FileNotFoundError:
    print("Erro: O arquivo não foi encontrado.")


# Etapa 4: Fechamento e cleanup com finally

arquivo = None

try:
    arquivo = open("acessos.log", "r", encoding="utf-8")
    conteudo = arquivo.read()
    print("\nLeitura realizada com sucesso.")

except FileNotFoundError:
    print("Erro: Arquivo não encontrado.")

except Exception as erro:
    print("Erro inesperado:", erro)

finally:
    if arquivo is not None:
        arquivo.close()
        print("Arquivo fechado com segurança.")
    else:
        print("Nenhum arquivo precisou ser fechado.")
```
